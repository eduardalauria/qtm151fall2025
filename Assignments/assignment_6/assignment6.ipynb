(b) Subsetting Data

<font size = "3" >

- Read in the dataset named "features.csv".
- Define a global variable called ```threshold``` that takes the <br>
value 2500
- Subset the data to get one dataset with weight $>$ threshold and <br>
another with weight $\le$ threshold. Use the ```.query()``` function.
- Compute the mean acceleration for each of these datasets, printing these values to the screen<br>

Note: Review how to define a global variable inside ```.query()```
 

````python
# Write your answer here:
df = pd.read_csv('/Users/eduardalauria/QTM 151/qtm151fall2025/Assignments/assignment_6/data/features.csv')

threshold = 2500

gretan_than_threshold = df.query("weight > @threshold")
smaller_than_threshold =  df.query("weight <= @threshold")

mean_acceleration_greater = gretan_than_threshold['acceleration'].mean()
print(f"Mean acceleration for weight > threshold: {mean_acceleration_greater}")

mean_acceleration_smaller = smaller_than_threshold['acceleration'].mean()
print(f"Mean acceleration for weight <= threshold: {mean_acceleration_smaller} ")
````

<font size = "3">

Read in the dataset named "wdi_urban.csv".

  This contains data from 1980 and 2020.

- **Create an array with the unique values of "year" using the ``Pandas`` function ``unique``**

- **Create a plot with a ...**
    - Histogram of "prop_urbanpopulation" (proportion of urban population) for each year
    - Set the histogram option `alpha = 0.5` for transparency
    - All the histograms should be overlapping on the same plot
    - Create a legend with the list of years
    - Label the axes and the title

**Hint**: The procedure will be similar (but not exactly the same!) as when we created a scatter plot of cars with different numbers of cylinders in Lecture 11

````python
# your answer here
wdi_urban_df = pd.read_csv('/Users/eduardalauria/QTM 151/qtm151fall2025/Assignments/assignment_6/data/wdi_urban.csv')
unique_years = wdi_urban_df['year'].unique()

for year in unique_years:
    subset = wdi_urban_df[wdi_urban_df['year'] == year]
    plt.hist(subset['prop_urbanpopulation'], alpha=0.5, label=str(year))

plt.xlabel("Proportion of Urban Population")
plt.ylabel("Frequency")
plt.title("Histogram of Urban Population Proportion (1980 vs 2020)")
plt.legend()
plt.show()
````
